### `print` frente a `return`
`print` **muestra** un valor en pantalla; `return` **devuelve** el valor a quien llamó la función para seguir usándolo. Una función sin `return` devuelve `None`.

```python
def iva_con_print(precio):
    print(precio * 0.15)

def iva_con_return(precio):
    return precio * 0.15

a = iva_con_print(100)    # muestra 15.0, pero a queda en None
b = iva_con_return(100)   # no muestra nada, b vale 15.0
print("a =", a, "| b =", b)
```

> 📌 `return` **termina** la función de inmediato. Una función puede tener varios `return`, por ejemplo uno en cada rama de un `if`.

```python
def categoria(saldo):
    if saldo >= 1000:
        return "Oro"
    if saldo >= 100:
        return "Plata"
    return "Básica"

for s in [1500, 350, 20]:
    print(s, "→", categoria(s))
```

## 3. Parámetros (15 min)

### Posicionales y por nombre
Los argumentos se asignan **por posición**, o bien **por nombre** (en cualquier orden). Esto último no existe en Java.

```python
def cuota_simple(monto, tasa_anual, meses):
    interes = monto * tasa_anual * meses / 12
    return (monto + interes) / meses

print(round(cuota_simple(1200, 0.12, 6), 2))                       # por posición
print(round(cuota_simple(meses=6, monto=1200, tasa_anual=0.12), 2))  # por nombre
```

### Valores por defecto
Un parámetro con valor por defecto es **opcional**. En Java esto se resuelve con sobrecarga de métodos.

```python
def calcular_total(precio, cantidad=1, tasa_iva=0.15):
    subtotal = precio * cantidad
    return round(subtotal * (1 + tasa_iva), 2)

print(calcular_total(10))                  # 1 unidad con IVA 15 %
print(calcular_total(10, 3))               # 3 unidades
print(calcular_total(10, tasa_iva=0))      # sin IVA, cantidad por defecto
```

> ⚠️ Los parámetros con valor por defecto van **al final**: `def f(a=1, b)` produce `SyntaxError`.
>
> ⚠️ **Nunca** use una lista o un diccionario como valor por defecto (`def f(lista=[])`): se crea una sola vez y se comparte entre llamadas. Use `None` y créela dentro de la función.

```python
def agregar_mal(valor, lista=[]):
    lista.append(valor)
    return lista

print(agregar_mal(1))
print(agregar_mal(2))   # ¡sorpresa! [1, 2]

def agregar_bien(valor, lista=None):
    if lista is None:
        lista = []
    lista.append(valor)
    return lista

print(agregar_bien(1))
print(agregar_bien(2))  # [2]
```

## 4. Devolver varios valores y `None` (10 min)

En Java, para devolver varios datos hay que crear una clase o usar un arreglo. En Python, `return a, b` devuelve una **tupla**, que se desempaqueta al recibirla.

```python
def estadisticas(notas):
    promedio = sum(notas) / len(notas)
    return promedio, max(notas), min(notas)

prom, mayor, menor = estadisticas([8.5, 6.0, 9.75, 7.0])
print(f"Promedio: {prom:.2f} | Mayor: {mayor} | Menor: {menor}")

resultado = estadisticas([10, 9])
print(resultado, type(resultado))
```

```python
def buscar_socio(socios, cedula):
    for s in socios:
        if s["cedula"] == cedula:
            return s
    return None      # no se encontró

socios = [{"cedula": "0102030405", "nombre": "Ana"}, {"cedula": "0912345678", "nombre": "Luis"}]
encontrado = buscar_socio(socios, "0912345678")
print(encontrado["nombre"] if encontrado is not None else "No registrado")
print(buscar_socio(socios, "0000000000"))
```

## 5. Ámbito: variables locales y globales (10 min)

- Una variable creada **dentro** de una función es **local**: solo existe mientras la función se ejecuta.
- Una variable creada **fuera** de las funciones es **global**: se puede **leer** dentro de ellas.
- Para **modificar** una global dentro de una función hay que declararla con `global`, pero es mejor evitarlo: reciba el valor como parámetro y devuelva el resultado.

```python
TASA_IVA = 0.15          # global (constante, por convención en MAYÚSCULAS)

def precio_final(precio):
    iva = precio * TASA_IVA   # iva es local; TASA_IVA se puede leer
    return precio + iva

print(precio_final(20))
# print(iva)  → NameError: iva solo existe dentro de la función
```

```python
saldo = 100

def depositar_mal(monto):
    global saldo          # funciona, pero esconde un cambio de estado
    saldo += monto

def depositar_bien(saldo_actual, monto):
    return saldo_actual + monto   # recibe y devuelve: fácil de probar

depositar_mal(50)
print(saldo)
saldo = depositar_bien(saldo, 50)
print(saldo)
```

## 7. Documentar: docstrings y anotaciones de tipo (10 min)

- El **docstring** es una cadena entre triples comillas en la primera línea de la función. Equivale al Javadoc y se consulta con `help()`.
- Las **anotaciones de tipo** (*type hints*) indican qué tipos se esperan. Python **no** las obliga al ejecutar, pero ayudan a leer el código y al editor a detectar errores.

```python
def calcular_cuota(monto: float, tasa_anual: float, meses: int) -> float:
    """Calcula la cuota fija mensual de un crédito con el sistema francés.

    monto: valor del préstamo en dólares.
    tasa_anual: tasa de interés anual en decimal (0.12 = 12 %).
    meses: número de cuotas.
    Devuelve la cuota mensual redondeada a 2 decimales.
    """
    i = tasa_anual / 12
    return round(monto * i / (1 - (1 + i) ** -meses), 2)

print(calcular_cuota(1200, 0.12, 6))
help(calcular_cuota)
```

## 8. Funciones `lambda` y funciones como argumento (10 min)

En Python las funciones son **valores**: se pueden guardar en variables y pasar a otras funciones. Una `lambda` es una función **anónima de una sola expresión**. Equivale a las expresiones lambda de Java: `x -> x * 2`.

```python
doble = lambda x: x * 2
print(doble(21))

def aplicar(funcion, valor):
    return funcion(valor)

print(aplicar(doble, 5))
print(aplicar(lambda x: x ** 2, 5))
```

```python
socios = [("Ana", 1520.75), ("Luis", 80.0), ("María", 640.3)]

# Ordenar por saldo (segundo elemento), de mayor a menor
for nombre, saldo in sorted(socios, key=lambda s: s[1], reverse=True):
    print(f"{nombre:<6}{saldo:>9,.2f}")

# map y filter: transforman y filtran con una función
saldos = [s[1] for s in socios]
print(list(map(lambda x: round(x * 1.02, 2), saldos)))
print(list(filter(lambda x: x >= 100, saldos)))
```

> 💡 Para transformar o filtrar listas, las **comprensiones** del tema 1.3 suelen ser más legibles que `map` y `filter`. La `lambda` se usa sobre todo como `key=` en `sorted`, `max` y `min`.

## 9. Caso práctico: el cajero de la cooperativa con funciones (15 min)

Reorganizamos el cajero del tema 1.3: cada operación es una función pequeña, con un solo propósito, que **recibe datos y devuelve resultados**. El menú solo coordina.

```python
def monto_valido(monto: float, maximo: float = 1000) -> bool:
    """True si el monto es positivo y no supera el máximo por operación."""
    return 0 < monto <= maximo

def depositar(saldo: float, monto: float) -> tuple:
    """Devuelve (nuevo_saldo, mensaje)."""
    if not monto_valido(monto):
        return saldo, "Monto inválido"
    return saldo + monto, f"Depósito de ${monto:.2f} realizado"

def retirar(saldo: float, monto: float) -> tuple:
    """Devuelve (nuevo_saldo, mensaje)."""
    if not monto_valido(monto):
        return saldo, "Monto inválido"
    if monto > saldo:
        return saldo, "Saldo insuficiente"
    return saldo - monto, f"Retiro de ${monto:.2f} realizado"

def mostrar_menu():
    print("\n1. Consultar saldo  2. Depositar  3. Retirar  0. Salir")

def cajero(saldo_inicial: float = 100.0):
    saldo = saldo_inicial
    while True:
        mostrar_menu()
        opcion = input("Opción: ")
        if opcion == "1":
            print(f"Saldo: ${saldo:,.2f}")
        elif opcion == "2":
            saldo, mensaje = depositar(saldo, float(input("Monto: ")))
            print(mensaje)
        elif opcion == "3":
            saldo, mensaje = retirar(saldo, float(input("Monto: ")))
            print(mensaje)
        elif opcion == "0":
            print("Gracias por usar el cajero")
            return saldo
        else:
            print("Opción no válida")

saldo_final = cajero()
print("Saldo al salir:", saldo_final)
```

### Ventaja: cada función se puede **probar** por separado
Con `assert` comprobamos que una función devuelve lo esperado. Si una prueba falla, Python muestra `AssertionError`.

```python
assert monto_valido(50) is True
assert monto_valido(-5) is False
assert monto_valido(5000) is False
assert depositar(100, 50) == (150, "Depósito de $50.00 realizado")
assert retirar(100, 500) == (100, "Saldo insuficiente")
print("Todas las pruebas pasaron ✔")
```

```python
# Ejercicio 1
def clasificar_edad(edad: int) -> str:
    """Devuelve 'menor', 'adulto' o 'adulto mayor' según la edad."""
    if edad < 18:
        return "menor"
    if edad < 65:
        return "adulto"
    return "adulto mayor"

assert clasificar_edad(17) == "menor"
assert clasificar_edad(18) == "adulto"
assert clasificar_edad(65) == "adulto mayor"
for e in [15, 30, 70, 18, 64]:
    print(e, "→", clasificar_edad(e))
```

```python
# Ejercicio 2
def calcular_factura(precio: float, cantidad: int = 1, descuento: float = 0, tasa_iva: float = 0.15) -> tuple:
    """Devuelve (subtotal, iva, total). El descuento se aplica antes del IVA."""
    subtotal = precio * cantidad * (1 - descuento)
    iva = subtotal * tasa_iva
    return round(subtotal, 2), round(iva, 2), round(subtotal + iva, 2)

assert calcular_factura(10) == (10.0, 1.5, 11.5)
assert calcular_factura(30, 3, descuento=0.10) == (81.0, 12.15, 93.15)
print(calcular_factura(30, cantidad=3))
print(calcular_factura(30, cantidad=3, descuento=0.10))
```

```python
# Ejercicio 3
def contar_letras(texto: str) -> dict:
    """Cuenta vocales y consonantes de un texto; ignora espacios, números y signos."""
    vocales = consonantes = 0
    for c in texto.lower():
        if c in "aeiouáéíóúü":
            vocales += 1
        elif c.isalpha():
            consonantes += 1
    return {"vocales": vocales, "consonantes": consonantes}

assert contar_letras("Hola") == {"vocales": 2, "consonantes": 2}
assert contar_letras("Año 2026, Perú!") == {"vocales": 4, "consonantes": 3}
print(contar_letras("Programación en Python"))
```

```python
# Ejercicio 4
def promedio(*notas: float):
    """Promedio de las notas recibidas, con 2 decimales; None si no hay notas."""
    if not notas:
        return None
    return round(sum(notas) / len(notas), 2)

def mejores(estudiantes: dict, n: int = 3) -> list:
    """Nombres de los n estudiantes con mejor nota."""
    ordenados = sorted(estudiantes, key=lambda nombre: estudiantes[nombre], reverse=True)
    return ordenados[:n]

assert promedio(8, 9, 10) == 9.0
assert promedio() is None
estudiantes = {"Ana": 9.5, "Luis": 7.25, "María": 8.75, "Pedro": 6.5, "Sofía": 10}
assert mejores(estudiantes) == ["Sofía", "Ana", "María"]
print(promedio(*estudiantes.values()))
print(mejores(estudiantes, n=2))
```

```python
# Ejercicio 5
def digito_verificador(cedula: str) -> int:
    """Calcula el dígito verificador esperado de una cédula ecuatoriana."""
    suma = 0
    for digito, coef in zip(cedula[:9], [2, 1, 2, 1, 2, 1, 2, 1, 2]):
        producto = int(digito) * coef
        if producto > 9:
            producto -= 9
        suma += producto
    verificador = 10 - suma % 10
    return 0 if verificador == 10 else verificador

def cedula_valida(cedula: str) -> bool:
    """True si la cédula tiene 10 dígitos, provincia 01-24 y verificador correcto."""
    if len(cedula) != 10 or not cedula.isdigit():
        return False
    if not 1 <= int(cedula[:2]) <= 24:
        return False
    return digito_verificador(cedula) == int(cedula[9])

assert cedula_valida("0104567896") is True
assert cedula_valida("0104567890") is False
assert cedula_valida("12345") is False
assert cedula_valida("9904567896") is False
for c in ["0104567896", "0104567890", "abc"]:
    print(c, "→", "válida" if cedula_valida(c) else "no válida")
```
