4. Agregue un nuevo vendedor `"Pedro"` con ventas `[200, 210, 190, 260]`.
5. Muestre la mejor venta de Ana con `max`.

````python
vendedores = ["Ana", "Luis", "María"]
ventas = [[120, 340, 210, 400],
          [80, 150, 300, 220],
          [500, 260, 180, 310]]
# Escriba su solución aquí
print ("============================== Solucion ==============================")
print("Primer ejercicio:")
print("Las ventas de Luis en la semana 3 fueron: ", ventas[1][2])
print("Segundo ejercicio:")
print("Las ventas totales de Maria fueron: ", sum(ventas[2]))
print("Tercer ejercicio:")
total_semana1= ventas[0][0]+ventas[1][0]+ventas[2][0]
print("Las ventas totales de la semana 1 fueron: ", total_semana1)
print("Cuarto ejercicio:")
vendedores.append("Pedro")
ventas.append([200, 210, 190, 260])
print("Los vendedores actualizados son: ", vendedores)
print ("Las ventas actualizadas son: ", ventas)
print("Quinto ejercicio:")
print("La mejor venta de Ana fue: ", max(ventas[0]))
````
