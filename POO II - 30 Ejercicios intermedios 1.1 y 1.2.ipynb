### Ejercicio 1 · Factura con descuento e IVA ★★

Pida el nombre de un producto, la cantidad y el precio unitario. Calcule:
- Subtotal = cantidad × precio
- Descuento del 10 % sobre el subtotal
- IVA del 15 % sobre (subtotal − descuento)
- Total

Muestre la factura con los valores **alineados a la derecha** en 10 espacios y 2 decimales:
```
Producto: Teclado
Subtotal:      90.00
Descuento:      9.00
IVA:           12.15
TOTAL:         93.15
```

````python
# Escriba su solución aquí
producto = input("Ingrese el nombre del producto: ")
cantidad = int(input("Ingrese la cantidad: "))
precio_unitario = float(input("Ingrese el precio unitario: "))
subtotal=cantidad*precio_unitario
descuento=subtotal-(subtotal*0.10)
iva=0.15*(subtotal-descuento)
total = subtotal - descuento + iva
print(f"Producto: {producto}")
print(f"Subtotal: {subtotal:>10.2f}")
print(f"Descuento: {descuento:>10.2f}")
print(f"IVA: {iva:>10.2f}")
print(f"Total: {total:>10.2f}")
````

### Ejercicio 16 · Limpiar un registro ★★

Un sistema exportó este registro con espacios y mayúsculas desordenadas:
`registro = "  0104567896 ,  pérez  ANA ,  ANA.PEREZ@MAIL.COM ,  150.50  "`

````python
registro = "  0104567896 ,  pérez  ANA ,  ANA.PEREZ@MAIL.COM ,  150.50  "
# Escriba su solución aquí
partes = registro.split(",")

cedula = partes[0].strip()
nombre = " ".join(partes[1].strip().split()).title()
correo = partes[2].strip().lower()
saldo = float(partes[3].strip())

datos = {
    "cedula": cedula,
    "nombre": nombre,
    "correo": correo,
    "saldo": saldo
}

print("Los datos ordenados son: ", datos)
````

```python
python_t = {"Ana", "Luis", "María", "Pedro", "Sofía"}
excel_t  = {"Luis", "Carla", "Pedro", "Diego"}
ia_t     = {"Ana", "Pedro", "Diego"}
```
Muestre (ordenado con `sorted`):
1. Quiénes están en Python **y** en Excel.
2. Quiénes están en los tres talleres.
3. Quiénes están en Python pero **no** en IA.
4. Quiénes están en Python **o** en Excel, pero no en ambos (`^`).
5. Si todos los de IA están también en Python (`issubset`).
6. El total de estudiantes distintos.

````python
python_t = {"Ana", "Luis", "María", "Pedro", "Sofía"}
excel_t = {"Luis", "Carla", "Pedro", "Diego"}
ia_t = {"Ana", "Pedro", "Diego"}
# Escriba su solución aquí
python_excel=sorted(python_t & excel_t)
tres_talleres=sorted(python_t & excel_t & ia_t)
python_no_ia=sorted(python_t - ia_t)
python_o_excel_no_ambos = sorted(python_t ^ excel_t)
ia_en_python = ia_t.issubset(python_t)
total_distintos=len(python_t | excel_t | ia_t)

print(f"En Python y Excel: {python_excel}")
print(f"En los tres talleres: {tres_talleres}")
print(f"En Python pero no en IA: {python_no_ia}")
print(f"En Python o Excel, pero no en ambos: {python_o_excel_no_ambos}")
print(f"Todos los de IA están en Python: {ia_en_python}")
print(f"Total de estudiantes distintos: {total_distintos}")
````

### Ejercicio 29 · Copia o referencia ★★

Ejecute y explique en comentarios cada resultado:

original = {"nombre": "Ana", "saldo": 100}
alias = original
copia = original.copy()

1. Muestre `original`, `alias` y `copia`.

3. Compare con `id()` las direcciones de memoria.

### Ejercicio 30 · Ficha de un socio a partir de texto ★★

Un archivo antiguo guarda cada socio en una sola línea:
`linea = "ana pérez|0104567896|ahorros=1520.75|plazo fijo=5000"`
1. Separe la línea por `"|"`.
2. Cree un diccionario con `nombre` (formato título), `cedula` y `cuentas`, que a su vez es un diccionario `{tipo: saldo}` (separe cada cuenta por `"="` y convierta el saldo a `float`).
3. Muestre la ficha con este formato:
```
Socio: Ana Pérez (0104567896)
Tipos de cuenta: ahorros, plazo fijo
Saldo total: $6,520.75
```

````python
linea = "ana pérez|0104567896|ahorros=1520.75|plazo fijo=5000"
# Escriba su solución aquí

partes=linea.split("|")
cuenta1 = partes[2].split("=")
cuenta2 = partes[3].split("=")
ficha = {
    "nombre": partes[0].title(),
    "cedula": partes[1],
    "cuentas": {
        cuenta1[0]: float(cuenta1[1]),
        cuenta2[0]: float(cuenta2[1])
    }
}
saldo_final=sum(ficha["cuentas"].values())

print(f"Socio: {ficha['nombre']} ({ficha['cedula']})")
print(f"Tipos de cuenta: {cuenta1[0]}, {cuenta2[0]}")
print(f"Saldo total: ${saldo_final:,.2f}")
````
