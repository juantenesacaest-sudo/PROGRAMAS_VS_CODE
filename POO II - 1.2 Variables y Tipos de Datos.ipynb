````python
maximo_java = 2147483647      # el guion bajo solo mejora la lectura
print(maximo_java + 465451651650000)           # en Java (int) daría -2147483648
print(maximo_java ** 100)
````

### 6.1 Listas: `list`
Java necesita `ArrayList<Double> saldos = new ArrayList<>();`. En Python basta con corchetes, y una lista puede mezclar tipos.

````python
saldos = [100.0, 250.5, 80.0]

saldos.append(500.0)        # agregar al final  (Java: add)
saldos.insert(0, 20.0)      # insertar en una posición
print(saldos)

print(saldos[1], saldos[-1])  # acceso por índice
saldos[0] = 25.0              # modificar (Java: set)
saldos.remove(80.0)           # eliminar por valor
print(saldos)

print(len(saldos), sum(saldos), max(saldos), min(saldos))
print(sorted(saldos))         # nueva lista ordenada
print(250.5 in saldos)        # Java: contains
````

````python
estudiantes = ["Lumbi", "Minchala", "Juela", "Frixon", "Anderson"]
estudiantes.append("Juan")
print(estudiantes)
estudiantes.insert(1, "Kevin")
print(estudiantes)
estudiantes.remove("Juela")
print(estudiantes)
print(estudiantes[4])
print("Tamaño:", len(estudiantes))
print(estudiantes[::-1])
print(sorted(estudiantes)) # Para ordenar la lista de estudiantes en orden alfabético
````

El *slicing* también funciona con listas:

````python
meses = ["ene", "feb", "mar", "abr", "may", "jun"]
print(meses[:3])     # primer trimestre
print(meses[3:])     # segundo trimestre
print(meses[::-1])   # invertida
````

### 6.3 Diccionarios: `dict`
Un diccionario guarda pares **clave → valor**, como un `HashMap<String, Object>` de Java.

````python
socio = {
    "cedula": "0102030405",
    "nombre": "Ana Pérez",
    "saldo": 1520.75,
    "activo": True,
}

print(socio["nombre"])              # Java: map.get("nombre")
socio["saldo"] += 100               # modificar
socio["telefono"] = "0991234567"    # agregar una clave nueva (Java: put)
print(socio.get("correo", "sin correo"))  # valor por defecto si no existe

print(socio.keys())
print("cedula" in socio)            # Java: containsKey
````

````python
estudiante = {
    "Nombre":"David Tenesaca",
    "Cédula":"0102030405",
    "Edad":18,
    "Direccion":"Av. Universitaria 123",
    "Lista_notas":[8.5, 9.0, 7.5, 8.0, 9.5, 10.0],
    "Correo":"juan@edu.edu.ec"
}
print(estudiante["Nombre"])
print(estudiante.keys())
print(estudiante["Lista_notas"])
print(estudiante["Lista_notas"][4])
print(estudiante.get("Direccion", "No tiene dirección"))
````

### 6.4 Conjuntos: `set`
Un conjunto **no admite repetidos**, como el `HashSet` de Java. Es útil para eliminar duplicados.

````python
tipos_cuenta = ["ahorros", "corriente", "ahorros", "plazo fijo", "ahorros"]
unicos = set(tipos_cuenta)
print(unicos, len(unicos))

cuentas_ana = {"ahorros", "plazo fijo"}
cuentas_luis = {"ahorros", "corriente"}
print(cuentas_ana & cuentas_luis)   # intersección: lo que tienen en común
print(cuentas_ana | cuentas_luis)   # unión
````

## 7. Mutabilidad y referencias (10 min)

| Inmutables (no cambian) | Mutables (sí cambian) |
|---|---|
| `int`, `float`, `bool`, `str`, `tuple` | `list`, `dict`, `set` |

Igual que con los objetos en Java, **una variable guarda una referencia**. Si dos variables apuntan a la misma lista, un cambio hecho con una se ve con la otra:

````python
cuentas_a = ["AH-001", "AH-002"]
cuentas_b = cuentas_a          # NO es una copia: es la misma lista
cuentas_b.append("AH-003")
print(cuentas_a)               # ¡también cambió!

copia = cuentas_a.copy()       # copia independiente
copia.append("AH-004")
print(cuentas_a, copia)
````

`==` compara el **contenido** (como `equals` en Java) e `is` compara si es **el mismo objeto** (como `==` con objetos en Java):

````python
x = [1, 2, 3]
y = [1, 2, 3]
print(x == y)   # True: mismo contenido
print(x is y)   # False: son dos listas distintas
````

## Soluciones de los ejercicios

````python
# Ejercicio 1
dato = "  0102030405 ; MARÍA josé torres ; 250.50  "
cedula, nombre, saldo = [p.strip() for p in dato.split(";")]
nombre = nombre.title()
saldo = float(saldo) + 49.50
print(nombre)
print(f"Saldo: ${saldo:.2f}")
print("Cédula válida:", cedula.isdigit() and len(cedula) == 10)
````

````python
# Ejercicio 2
depositos = [120.0, 45.5, 300.0, 45.5, 80.0]
depositos.append(200.0)
total = sum(depositos)
print(f"Total: ${total:.2f}")
print(f"Promedio: ${total / len(depositos):.2f}")
print("Mayor:", max(depositos), "| Menor:", min(depositos))
print("Montos distintos:", len(set(depositos)))
````
