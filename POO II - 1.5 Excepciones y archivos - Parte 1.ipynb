## 2. ¿Qué es una excepción? (15 min)

Una **excepción** es un error que ocurre **mientras el programa se ejecuta**. Si nadie la maneja, Python detiene el programa y muestra el **traceback**: la cadena de llamadas, la línea del error, el **tipo** de excepción y un mensaje.

> 📌 Se lee **de abajo hacia arriba**: la última línea dice qué pasó y la anterior, dónde.

| Excepción | Cuándo ocurre | Ejemplo |
|---|---|---|
| `ValueError` | El tipo es correcto pero el valor no sirve | `int("hola")` |
| `TypeError` | Operación con tipos incompatibles | `"5" + 3` |
| `ZeroDivisionError` | División para cero | `10 / 0` |
| `IndexError` | Índice fuera de rango | `[1, 2][5]` |
| `KeyError` | Clave que no existe en un diccionario | `{"a": 1}["b"]` |
| `NameError` | Variable que no existe | `print(saldoo)` |
| `FileNotFoundError` | Archivo que no existe | `open("x.txt")` |
| `AttributeError` | Método o atributo que no existe | `"hola".push("a")` |

Ejecute cada línea por separado (quite el `#`) y lea el mensaje:

```python
#int("hola")
#"5" + 3
#10 / 0
#[1, 2][5]
# {"a": 1}["b"]
print("Quite el # de una línea a la vez para ver cada excepción")
```

### Comparación con Java
| Java | Python |
|---|---|
| `try { … } catch (NumberFormatException e) { … }` | `try:` … `except ValueError as e:` |
| `finally { … }` | `finally:` |
| `throw new IllegalArgumentException("…")` | `raise ValueError("…")` |
| Excepciones *checked*: hay que declararlas con `throws` | **No existen**: ninguna se declara |
| `ArithmeticException` | `ZeroDivisionError` |
| `ArrayIndexOutOfBoundsException` | `IndexError` |
| `NullPointerException` | Lo más parecido: `AttributeError` sobre `None` |

## 3. `try` / `except` (20 min)

El código que **puede fallar** va en `try`. Si ocurre una excepción del tipo indicado, Python salta al `except` y el programa **continúa**.

```python
texto = "cien"
try:
    monto = float(texto)
    print("Monto:", monto)
except ValueError:
    print(f"'{texto}' no es un número válido")
print("El programa sigue funcionando")
```

```python
try:
    numero=int(input("Ingrese numero "))
    print(numero/0)
except ValueError as e:
    print("Error tipo", e)
except ZeroDivisionError:
    print("No se puede dividir para cero")
```

### Capturar varios tipos de error
Se pueden escribir varios `except`, uno por tipo, o agrupar tipos en una tupla. Con `as e` obtenemos el objeto de la excepción y su mensaje.

```python
def promedio_por_persona(total, personas):
    try:
        return total / personas
    except ZeroDivisionError:
        print("Error: el número de personas no puede ser 0")
    except TypeError as e:
        print("Error de tipo:", e)
    except ValueError as e:
        input("No puede ser un texto", e)
    return None

print("1. ",promedio_por_persona(100, 4))
print("2. ",promedio_por_persona(100, 0))
print("3. ",promedio_por_persona(100, "4"))
print("4. ",promedio_por_persona(input("Ingrese el total: "), 4))
```

```python
datos = {"Ana": [9, 8], "Luis": []}

for nombre in ["Ana", "Luis", "Pedro"]:
    try:
        notas = datos[nombre]
        print(nombre, "→", sum(notas) / len(notas))
    except (KeyError, ZeroDivisionError) as e:
        print(nombre, "→ sin datos:", type(e).__name__)
```

> ⚠️ El **orden** de los `except` importa: Python usa el **primero** que coincide. Ponga primero los más específicos y al final los más generales (`Exception`).

## 4. `else` y `finally` (10 min)

| Bloque | Se ejecuta… |
|---|---|
| `try` | siempre: es el código que puede fallar |
| `except` | solo si hubo una excepción de ese tipo |
| `else` | solo si **no** hubo ninguna excepción |
| `finally` | **siempre**, haya error o no (cerrar archivos, liberar recursos) |

```python
def convertir(texto):
    try:
        valor = float(texto)
    except ValueError:
        print(f"  '{texto}': no es un número")
    else:
        print(f"  '{texto}': convertido a {valor}")
    finally:
        print("  (fin del intento)")

for t in ["12.5", "abc"]:
    convertir(t)
```

```python
def operaciones(a, b):
    try:
        suma=int(a)+int(b)
        ope=a/b
    except(ValueError, TypeError, ZeroDivisionError) as e:
        print("Error de tipo: ", e)
    else:
        print("Se ha ejecutado correctamente")
        print(suma)
        print(ope)
    finally:
        print("Fin de la ejecución")

operaciones(10,5)
```

## 5. Validar la entrada del usuario (15 min)

Combinamos `while True` (tema 1.3), una función (tema 1.4) y `try`/`except`: se pide el dato **hasta que sea válido**. Pruebe escribiendo letras, un número negativo y luego un número correcto.

```python
def leer_monto(mensaje="Monto: "):
    """Pide un monto hasta que el usuario escriba un número positivo."""
    while True:
        texto = input(mensaje)
        try:
            monto = float(texto)
        except ValueError:
            print("Escriba un número, por ejemplo 25.50")
            continue
        if monto <= 0:
            print("El monto debe ser mayor que cero")
            continue
        return monto

monto = leer_monto()
print(f"Monto aceptado: ${monto:.2f}")
```

## 6. Lanzar excepciones con `raise` (15 min)

Una función no debería imprimir el error y seguir como si nada: debe **avisar** a quien la llamó. Con `raise` se lanza una excepción; quien llama decide cómo manejarla. Equivale a `throw` en Java.

```python
def retirar(saldo, monto):
    """Devuelve el nuevo saldo. Lanza ValueError si el retiro no es posible."""
    if monto <= 0:
        raise ValueError("El monto debe ser positivo")
    if monto > saldo:
        raise ValueError(f"Saldo insuficiente: tiene ${saldo:.2f}")
    return saldo - monto

saldo = 100.0
for monto in [30, -5, 500]:
    try:
        saldo = retirar(saldo, monto)
        print(f"Retiro de ${monto} → saldo ${saldo:.2f}")
    except ValueError as e:
        print(f"Retiro de ${monto} rechazado: {e}")
```

> 💡 Ventaja frente a devolver `None` o un mensaje (como en el tema 1.4): el error **no se puede ignorar sin querer**. Si nadie lo captura, el programa se detiene y muestra exactamente qué pasó.

## 8. Buenas prácticas (5 min)

| ✅ Hacer | ❌ Evitar |
|---|---|
| Capturar el tipo **específico** (`except ValueError`) | `except:` vacío, que atrapa **todo**, incluso errores de programación |
| Poner en `try` **solo** las líneas que pueden fallar | Un `try` que envuelve 50 líneas |
| Mostrar un mensaje útil o volver a lanzar | `except: pass`, que esconde el error en silencio |
| Validar con `if` lo que se puede prever | Usar excepciones para el flujo normal del programa |

```python
# ❌ Mal: el except vacío oculta un error de escritura (saldoo no existe)
saldo = 100
try:
    print(saldoo + 50)
except:
    print("No se pudo calcular")   # ¿por qué? nunca lo sabremos

# ✅ Bien: capturamos solo lo que esperamos; el NameError sí se vería
try:
    total = saldo + float("50")
    print("Total:", total)
except ValueError as e:
    print("Monto inválido:", e)
```

### Ejercicio 1 · Leer un entero en un rango
Escriba `leer_entero(mensaje, minimo, maximo)` que pida un número entero con `input()` hasta que sea válido y esté entre `minimo` y `maximo`. Use `try`/`except` y muestre un mensaje distinto para cada error. Úsela para pedir una nota de 0 a 10.

```python
# Escriba su solución aquí
def leer_entero(mensaje, minimo, maximo):
    while True:
        try:
            numero = int(input(mensaje))

            if minimo <= numero <= maximo:
                return numero

            print(f"Ingrese un número entre {minimo} y {maximo}.")
        except ValueError:
            print("Error: debe ingresar un número entero.")

numero = leer_entero("Ingrese un número de 0 a 10: ", 0, 10)
print("Número ingresado:", numero)
```

### Ejercicio 2 · División segura
Escriba `dividir(a, b)` que devuelva `a / b` redondeado a 2 decimales. Capture `ZeroDivisionError` y `TypeError`, muestre un mensaje para cada uno y devuelva `None`. Use `else` para mostrar `"División correcta"` y `finally` para mostrar `"Operación terminada"`. Pruébela con `(10, 4)`, `(10, 0)` y `(10, "2")`.

```python
# Escriba su solución aquí
def dividir(a, b):
    try:
        resultado = round(a / b, 2)
    except ZeroDivisionError:
        print("Error: no se puede dividir entre cero.")
        return None
    except TypeError:
        print("Error: los valores deben ser números.")
        return None
    else:
        print("División correcta")
        return resultado
    finally:
        print("Operación terminada")


print(dividir(10, 4))
print(dividir(10, 0))
print(dividir(10, "2"))
```

### Ejercicio 3 · Convertir una lista de textos
Escriba `convertir_numeros(textos)` que reciba una lista como `["12", "3.5", "abc", "", "-7"]` y devuelva una tupla `(numeros, invalidos)`: la lista de los que se pudieron convertir a `float` y la lista de los que no.

```python
# Escriba su solución aquí
def convertir_numeros(textos):
    numeros = []
    invalidos = []

    for texto in textos:
        try:
            numeros.append(float(texto))
        except ValueError:
            invalidos.append(texto)

    return numeros, invalidos

resultado = convertir_numeros(["12", "3.5", "abc", "", "-7"])
print(resultado)
```

### Ejercicio 4 · Edad válida con `raise`
Escriba `validar_edad(edad)` que:
- lance `TypeError` si `edad` no es un `int` (use `isinstance`);
- lance `ValueError` si es negativa o mayor que 120;
- devuelva `True` si es válida.

Pruébela dentro de un `try` con los valores `25`, `-3`, `150` y `"20"`, mostrando el tipo de excepción y su mensaje.

```python
# Escriba su solución aquí
def validar_edad(edad):
    if not isinstance(edad, int):
        raise TypeError("La edad debe ser un entero.")

    if edad < 0 or edad > 120:
        raise ValueError("La edad debe estar entre 0 y 120.")

    return True
for edad in [25, -3, 150, "20"]:
    try:
        print(edad, "→", validar_edad(edad))
    except (TypeError, ValueError) as error:
        print(edad, "→", type(error).__name__ + ":", error)
```

### Ejercicio 5 · Transferencia entre cuentas
Cree la excepción `CuentaNoExisteError`. Escriba `transferir(cuentas, origen, destino, monto)`, donde `cuentas` es un diccionario `{numero: saldo}`. La función debe:
- lanzar `CuentaNoExisteError` si alguna cuenta no existe;
- lanzar `ValueError` si el monto no es positivo o si supera el saldo de origen;
- mover el dinero si todo es correcto.

Pruébela con `cuentas = {"AH-001": 500.0, "AH-002": 80.0}` y varias transferencias, algunas válidas y otras no.

```python
# Escriba su solución aquí
def transferir(cuentas, origen, destino, monto):
    if origen not in cuentas or destino not in cuentas:
        raise ValueError("La cuenta de origen o destino no existe.")

    if monto <= 0:
        raise ValueError("El monto debe ser positivo.")

    if monto > cuentas[origen]:
        raise ValueError("Saldo insuficiente en la cuenta de origen.")

    cuentas[origen] -= monto
    cuentas[destino] += monto


cuentas = {"AH-001": 500.0, "AH-002": 80.0}

try:
    transferir(cuentas, "AH-001", "AH-002", 100.0)
    print("Transferencia realizada:", cuentas)
except ValueError as error:
    print("Error:", error)
```
