## Ejercicio 11 – Validar contraseña

Pida una contraseña y verifique que cumpla todas estas reglas: tener al menos 8 caracteres, al menos un dígito y al menos una letra mayúscula. Recorra la cadena con `for` (o use `any`). Muestre *Contraseña segura* o la lista de reglas que no cumple.

Escriba su código Python completo en el cuadro de respuesta. Respete la sangría de 4 espacios.

````python
# Escriba su solución aquí
print(30*"=", "EJERCICIO 11")
clave = input("Contraseña: ")
errores = []
largo = len(clave)

if largo < 8:
    errores.append("Debe tener al menos 8 caracteres")

tiene_digito = False
tiene_mayusculas = False
for i in clave:
    if i.isdigit():
        tiene_digito = True
    if i.isupper():
        tiene_mayusculas = True
if not tiene_digito:
    errores.append("Debe tener almenos un dígito")
if not tiene_mayusculas:
    errores.append("Debe tener al menos una mayuscula")
if len(errores) == 0:
    print("Contraseña segura")
    print(clave)
else:
    print("La contraseña no cumple con los requisitos")
    for o in errores:
        print("-", o)
````
